¿Por qué es importante?

#### Convertir variables categóricas a binarias

```python
# Función para mapear y manejar Si, No y NaN
def map_rain(value):
    if pd.isna(value):
        return 0
    elif value == 'Yes':
        return 1
    elif value == 'No':
        return 0
    else:
        return 0

# Aplicar la función a las columnas
df['RainToday'] = df['RainToday'].apply(map_rain)
df['RainTomorrow'] = df['RainTomorrow'].apply(map_rain)
```

```python
print(df['RainToday'].unique())
print(df['RainTomorrow'].unique())
```

**Descripción:**
Este paso transforma las variables RainToday y RainTomorrow, originalmente con valores tipo string ("Yes" y "No"), a un formato numérico binario (1 y 0). Esta codificación es crucial para permitir su uso en modelos de predicción, ya que muchos algoritmos de machine learning requieren variables numéricas como entrada.

**Motivo de la transformación:**
Facilita el análisis estadístico, la correlación y el entrenamiento de modelos predictivos, especialmente cuando RainTomorrow será utilizada como variable objetivo (target) para predecir si lloverá al día siguiente.

#### Imputación de datos faltantes con Mediana

_Variables numericas que requieren imputación por cantidad de valores faltantes_

```python
# Variables numéricas a imputar
variables_numericas = ['Sunshine', 'Evaporation', 'Pressure9am', 'Pressure3pm', 'WindGustSpeed', 'Humidity3pm', 'Temp3pm', 'WindSpeed3pm', 'Humidity9am', 'Rainfall', 'WindSpeed9am', 'Temp9am', 'MinTemp', 'MaxTemp']
```

_Datos antes de la imputación_

```python
# Resumen antes de la imputación
print("Valores nulos antes de la interpolación:")
print(df[variables_numericas].isnull().sum())
```

##### Imputación de datos por media

```python
# Imputar con la media
for columna in variables_numericas:
    media = df[columna].mean()
    df[columna].fillna(media, inplace=True)

# Verificar valores nulos después de la imputación
print(df[variables_numericas].isnull().sum())
```

✅ **Datos Listos para el Análisis:**
- Ahora, las variables numéricas están completas y listas para ser utilizadas en análisis o modelado.
